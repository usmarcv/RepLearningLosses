```python
# Load Torch Data
test_embeds = torch.load('save/linear/path_models/path_lr_0.0001_bsz_32_SINCERE/embeds.pth', weights_only=False)
test_labels = torch.load('save/linear/path_models/path_lr_0.0001_bsz_32_SINCERE/labels.pth', weights_only=False)
print(test_embeds.shape, test_labels.shape)
```

#Observando os espaços de características

##Projeção Multidimensional (t-SNE)

Documentation: https://scikit-learn.org/stable/modules/generated/sklearn.manifold.TSNE.html

```python
from sklearn.manifold import TSNE
```

```python
X = np.array(test_embeds)
X_embedded = TSNE(n_components=2, learning_rate='auto', init='random', perplexity=25).fit_transform(X)
X_embedded.shape
```

```python
df_y  = pd.DataFrame(test_labels)
df_X  = pd.DataFrame(X_embedded)
df_Xy = pd.concat([df_X, df_y], axis=1)
df_Xy.columns = ["x", "y", "Class"]
df_Xy[0:5]
```

```python
import plotly.express as px
fig = px.scatter(df_Xy, x="x", y="y", color="Class", height=600, width=1200)
fig.show()
```

##Projeção Multidimensional (PCA)

Documentation: https://scikit-learn.org/dev/modules/generated/sklearn.decomposition.PCA.html

```python
from sklearn.decomposition import PCA
```

```python
X = np.array(test_embeds)
X_embedded = PCA(n_components=2).fit_transform(X)
X_embedded.shape
```

```python
df_y  = pd.DataFrame(test_labels)
df_X  = pd.DataFrame(X_embedded)
df_Xy = pd.concat([df_X, df_y], axis=1)
df_Xy.columns = ["x", "y", "Class"]
df_Xy[0:5]
```

```python
import plotly.express as px
fig = px.scatter(df_Xy, x="x", y="y", color="Class", height=600, width=1200)
fig.show()
```

##Projeção Multidimensional (MDS)

Documentation: https://scikit-learn.org/dev/modules/generated/sklearn.manifold.MDS.html

```python
from sklearn.manifold import MDS
from sklearn.metrics.pairwise import cosine_distances
```

```python
X = np.array(test_embeds).astype(np.float64)
X_embedded = MDS(n_components=2, dissimilarity='euclidean', normalized_stress='auto').fit_transform(X)
# dissimilarities = cosine_distances(X)
# X_embedded = MDS(n_components=2, dissimilarity='precomputed', normalized_stress='auto').fit_transform(dissimilarities)
X_embedded.shape
```

```python
df_y  = pd.DataFrame(test_labels)
df_X  = pd.DataFrame(X_embedded)
df_Xy = pd.concat([df_X, df_y], axis=1)
df_Xy.columns = ["x", "y", "Class"]
df_Xy[0:5]
```

```python
import plotly.express as px
fig = px.scatter(df_Xy, x="x", y="y", color="Class", height=600, width=1200)
fig.show()
```
