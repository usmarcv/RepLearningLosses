## Testando baixar vários modelos DINOv1

```python
import os
import requests
import argparse
from tqdm import tqdm

# URLs dos modelos DINOv1 no github
DINO_MODELS = {
    "dino_vit_small": "https://dl.fbaipublicfiles.com/dino/dino_deitsmall16_pretrain.pth",
    "dino_vit_base": "https://dl.fbaipublicfiles.com/dino/dino_vitbase16_pretrain.pth",
    "dino_vit_small_8": "https://dl.fbaipublicfiles.com/dino/dino_deitsmall8_pretrain.pth",
    "dino_vit_base_8": "https://dl.fbaipublicfiles.com/dino/dino_vitbase8_pretrain.pth",
    "dino_resnet50": "https://dl.fbaipublicfiles.com/dino/dino_resnet50_pretrain.pth",
}

def download_checkpoint(model_name, save_dir="checkpoints"):
    """Baixa um checkpoint do DINO e salva localmente."""
    if model_name not in DINO_MODELS:
        print(f"Modelo '{model_name}' não encontrado. Escolha entre: {list(DINO_MODELS.keys())}")
        return

    url = DINO_MODELS[model_name]
    filename = os.path.join(save_dir, os.path.basename(url))

    os.makedirs(save_dir, exist_ok=True)

    response = requests.get(url, stream=True)
    total_size = int(response.headers.get("content-length", 0))

    with open(filename, "wb") as file, tqdm(
        desc=f"Baixando {model_name}",
        total=total_size,
        unit="B",
        unit_scale=True,
        unit_divisor=1024,
    ) as bar:
        for chunk in response.iter_content(chunk_size=1024):
            file.write(chunk)
            bar.update(len(chunk))

    print(f"Download concluído: {filename}")

if __name__ == "__main__":
    parser = argparse.ArgumentParser(description="Baixar checkpoints do DINOv1.")
    parser.add_argument(
        "model_name",
        type=str,
        choices=DINO_MODELS.keys(),
        help="Nome do modelo para baixar."
    )
    parser.add_argument(
        "--save_dir",
        type=str,
        default="checkpoints",
        help="Diretório onde salvar o modelo."
    )
    args = parser.parse_args()

    download_checkpoint(args.model_name, args.save_dir)
```
